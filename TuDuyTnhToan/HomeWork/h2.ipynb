Bài 5: Viết chương trình giải phương trình bậc 2 (ax2 + bx + c = 0)

```python
from math import * 
a, b, c = map(int,input().split())
d = b**2 - 4*a*c
if a == 0 :
    if b == 0 and c == 0 : print('Pt vo so nghiem')
    elif b == 0 and c!= 0 : print('Pt vo nghiem')
else :
    if d >= 0 :
        x1 = (-b - sqrt(d)) / (2 * a)
        x2 = (-b + sqrt(d)) / (2 * a)
        if x1 == x2 : print(f'Pt co nghiem kep : x = {x1}')
        else : print(f'Pt co nghiem : x1 = {x1}, x2 = {x2}')
    else : print('Pt vo nghiem')
```

Bài 8: Giải hệ phương trình tuyến tính:
##### ax + by = m 
##### cx + dy = n

```python
a, b, m, c, d, n = map(int, input().split())
if a == c == 0 or b == d == 0 :
    if a == b == c == d == 0: print('Pt vo nghiem')
    else :print('Pt vo so ngiem')
else :
    x = (m -n * b / d) / (a - c * b / d)
    y = (m -n * a / c) / (b - d * a / c)
print(f'Hpt co nghiem la : (x, y) = ({x}, {y})')
```

Bài 9: Nhập vào thời gian 1 công việc nào đó là x giây. Hãy chuyển đổi và viết ra màn hình số thời gian trên dưới dạng bằng bao nhiêu giờ, bao nhiêu phút, bao nhiêu giây. 

```python
t = int(input())
h = t // 3600
m = (t - h * 3600) // 60
s = t - h * 3600 - m * 60
# print(f'{h} hours, {m} minutes, {s} seconds')
if h == 1 : 
    print(f'{h} hour,', end = ' ')
else :
    print(f'{h} hours,', end = ' ')
if m == 1 : 
    print(f'{m} minute,', end = ' ')
else :
    print(f'{m} minutes,', end = ' ')
if s == 1 : 
    print(f'{s} second')
else :
    print(f'{s} seconds')
```

Bài 10: Nhập vào tâm và bán kính của một đường tròn. Sau đó nhập vàomộtđiểm A(x, y) bất kì và kiểm tra xem nó có thuộc đường tròn hay không?
