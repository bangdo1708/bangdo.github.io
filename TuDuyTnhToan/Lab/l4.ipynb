### W4A6
"[SumPrimesInRange]
Nhập hai số nguyên dương a, b (a ≤ b). Hãy tính tổng các số nguyên tố trong đoạn
[a, b]. Mỗi dòng là một bộ test (a b)."

```python
from math import * 
def nt(a):
    if a < 2 : 
        return 0
    for i in range (2, int(sqrt(a) + 1)) :
        if a % i == 0 :
            return 0 
    return 1

if __name__ == '__main__' :
    a, b = map(int, input().split())
    s = 0
    for i in range(a, b+1) :
        if nt(i) == 1 :
            s += i
    print(s)
```

### W4A8
ReverseAddPalindrome]
Nhập số nguyên dương n. Lặp lại phép toán n = n + reverse(n) cho đến khi n là số
đối xứng (palindrome). In ra số bước và giá trị palindrome thu được. 


```python
from math import * 
def rv(a):
    s = 0
    while a > 0 :
        s = s * 10 + a % 10
        a //= 10
    return s
if __name__ == '__main__':
    a = int(input())
    s = 0
    while a != rv(a):
        a = a + rv(a)
        s += 1
    print (s, a)
```

### W4A9
[PerfectSquaresDistinctDigits]
Nhập vào số nguyên dương n. In ra tất cả các số chính phương ≤ n có các chữ số
đôi một khác nhau (in theo thứ tự tăng dần, cùng hàng, cách nhau bởi một dấu
cách). Nếu không có, in ra (no number).


```python
from math import * 
def t(a):
    s = str(a)
    if len (s) == len(set(s)) : return 1
    return 0
if __name__ == '__main__':
    n = int(input())
    if n <= 0 :
        print('No number')
    else :
        for i in range(1, int(sqrt(n) + 1)):
            if i ** 2 <= n and t(i**2) == 1:
                print(i**2, end = ' ')
```

### W4A10
"[LongestCollatzUpToN]
Nhập vào số nguyên dương n. Tìm số x trong [1, n] có độ dài dãy Collatz lớn nhất
và in ra x cùng độ dài dãy tương ứng (x L). Nếu có nhiều số cùng độ dài, chọn số
nhỏ nhất."

```python
from math import * 
def col(a):
    s = 1
    while a != 1 :
        if a % 2 == 0 : a /= 2
        else : a = a * 3 + 1
        s += 1 
    return s
if __name__ == '__main__':
    n = int(input())
    m = 1
    x = 1
    for i in range(1, n + 1):
        if m == col(i) :
            x = min(x, i)
        else :
            m = max(m, col(i))
            x = i
    print(x, m)
```

Hai số nguyên dương tạo thành một cặp số thân thiết khi chúng tuân theo quy luật:
Số này bằng tổng tất cả các ước của số kia (trừ chính nó) và ngược lại.
Ví dụ 220 và 284 là một cặp số thân thiết
220 ngoài chính nó ra, nó còn có 11 ước số khác là: 1, 2, 4, 5, 10, 11, 20, 44, 55,
110. Tổng của 11 ước này cho chúng ta 284
284 ngoài bản thân nó, còn có 5 ước số khác là: 1, 2, 4, 71, 142. Tổng của chúng
vừa đúng bằng 220
Nhập hai số từ bàn phím và kiểm tra xem chúng có phải là cặp số thân thiết không.
Nếu có, in ra màn hình "true", ngược lại, in ra "false"

```python
from math import *
def sum(a):
    s = 0
    for i in range (1, int(sqrt(a) + 1)):
        if a % i == 0 :
            s += i
        if a % int(a / i) == 0 and i != int(a / i) != a :
            s  += int(a / i)
    return s
if __name__ == '__main__':
    a, b = map(int, input().split())
    if sum(a) == b and sum(b) == a : print('True')
    else : print('False')
    # print(sum(a), sum(b), sep = '\n')
```

Sau khi nhập xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập 


```python
from math import *
first = True 
while True :
    n = int(input())
    if n == -1 : break
    if first :
        mx = mn = n
        first = False
    else : 
        mx = max (mx, n)
        mn = min (mn, n)
if first : 
    print('Invalid')
else : 
    print(f'max = {mx}, min = {mn}')
```

W4A26
[Fibonaci1]
Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ
bằng tổng hai số trước đó
Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt
quá A 


```python
from math import *
def fib(a):
    if a == 1 or a == 2 : return 1
    else : return fib(a - 1) + fib(a - 2)
if __name__=='__main__':
    n = int(input())
    print(fib(n))
```
